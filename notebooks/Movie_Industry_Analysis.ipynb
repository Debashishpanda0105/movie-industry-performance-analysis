## 3.4 Converting `released` to Datetime

The `released` column contains movie release dates, but it is currently
stored as a string.

For time-based analysis, the column should be converted to Pandas
datetime format.

We will use `pd.to_datetime()` with `errors="coerce"`.

If Pandas cannot interpret a value as a valid date, it will be converted
to `NaT` instead of stopping the entire operation.

````python
# Create a fresh cleaning copy
df_clean = df.copy()

# Remove country information from the released column
df_clean["released"] = (
    df_clean["released"]
    .str.replace(r"\s*\(.*\)", "", regex=True)
)

# Convert released column to datetime
df_clean["released"] = pd.to_datetime(
    df_clean["released"],
    errors="coerce"
)

# Check the result
df_clean["released"].head()
````

````python
# 3.5 Check Missing Dates After Conversion
df_clean["released"].isna().sum()
````

### Checking Leading and Trailing Spaces

Extra spaces can cause two values that look identical to be treated as
different categories.

For example:

`United States`

and

` United States `

would be treated as different strings.

We will check for unnecessary whitespace before standardizing the columns.

````python
for col in text_columns:
    spaces = df_clean[col].astype("string").str.strip().ne(
        df_clean[col].astype("string")
    ).sum()
    
    print(f"{col}: {spaces} values with extra spaces")
````

### Checking Empty Strings

Missing data is not always represented by `NaN`.

Sometimes missing values are stored as empty strings such as:

`""`

Therefore, we will check whether any text columns contain empty strings.

````python
for col in text_columns:
    empty_count = (
        df_clean[col]
        .astype("string")
        .str.strip()
        .eq("")
        .sum()
    )
    
    print(f"{col}: {empty_count} empty strings")
````

### Standardizing Whitespace

We will remove unnecessary leading and trailing spaces from text columns.

This improves consistency and prevents duplicate categories caused only
by whitespace differences.

````python
for col in text_columns:
    df_clean[col] = (
        df_clean[col]
        .astype("string")
        .str.strip()
    )
````

## 3.9 Handling Missing Values

Missing values must be handled carefully because different columns have
different meanings and business importance.

We will not use a single method such as `dropna()` for the entire dataset.

### Missing Value Strategy

| Column | Missing Values | Strategy |
|---|---:|---|
| `rating` | 77 | Replace with `"Unknown"` |
| `released` | 2 | Keep as `NaT` |
| `score` | 3 | Keep as missing |
| `votes` | 3 | Keep as missing |
| `writer` | 3 | Replace with `"Unknown"` |
| `star` | 1 | Replace with `"Unknown"` |
| `country` | 3 | Replace with `"Unknown"` |
| `company` | 17 | Replace with `"Unknown"` |
| `runtime` | 4 | Keep as missing |
| `budget` | 2,171 | Keep as missing |
| `gross` | 189 | Keep as missing |

### Why?

Categorical columns such as `rating`, `writer`, and `company` can use
`"Unknown"` because the missing value means that the information is
unavailable.

Numerical columns such as `score`, `votes`, `runtime`, `budget`, and
`gross` should not automatically be replaced with arbitrary values.

This prevents artificial data from being introduced into the analysis.

````python
# Replace missing categorical values with "Unknown"

categorical_missing = [
    "rating",
    "writer",
    "star",
    "country",
    "company"
]

for col in categorical_missing:
    df_clean[col] = df_clean[col].fillna("Unknown")
````

````python
df_clean[categorical_missing].isna().sum()
````

## 4.7 Estimated Profit

Estimated profit is calculated as:

Gross Revenue − Production Budget

This metric provides a simple measure of the financial outcome of a movie.

If either `budget` or `gross` is missing, profit will remain missing.

````python
df_clean["profit"] = (
    df_clean["gross"] - df_clean["budget"]
)
````

## 4.8 Return on Investment (ROI)

Return on Investment measures the financial return relative to the
production budget.

ROI is calculated using:

ROI = (Profit / Budget) × 100

A higher ROI indicates that a movie generated more revenue relative to
its production cost.

ROI will only be calculated when both `budget` and `gross` are available.

````python
df_clean["roi"] = (
    df_clean["profit"] / df_clean["budget"]
) * 100
````

````python
df_clean[
    ["name", "budget", "gross", "profit", "roi"]
].head(10)
````

## 4.9 Profitability Status

We will classify movies into three categories:

| Condition | Status |
|---|---|
| Profit > 0 | Profitable |
| Profit < 0 | Loss |
| Profit is missing | Unknown |

This feature will make profitability analysis easier.

````python
import numpy as np

df_clean["profit_status"] = np.select(
    [
        df_clean["profit"] > 0,
        df_clean["profit"] < 0
    ],
    [
        "Profitable",
        "Loss"
    ],
    default="Unknown"
)
````

## Business Question

Which movies generated the highest return relative to their
production budget?

## Objective

We will rank movies based on the `roi` column and identify
the Top 10 movies with the highest ROI.

## ROI Formula

ROI (%) = ((Gross Revenue - Budget) / Budget) × 100

Or:

ROI (%) = (Profit / Budget) × 100

## Why ROI?

Profit tells us the absolute amount of money earned.

ROI tells us how efficiently the invested money was converted
into profit.

Therefore, ROI is useful for comparing movies with very
different production budgets.

## Important Note

Movies with very small budgets can have extremely high ROI.
Therefore, ROI should always be interpreted together with
absolute profit and budget.

````python
roi_movies = (
    df_clean[
        ["name", "budget", "gross", "profit", "roi"]
    ]
    .dropna(subset=["roi"])
    .sort_values("roi", ascending=False)
)

roi_movies.head(10)
````

## Profit vs ROI

Profit and ROI should not be treated as interchangeable metrics.

### Example

Movie A:

Budget = $200M
Profit = $300M
ROI = 150%

Movie B:

Budget = $5M
Profit = $100M
ROI = 2000%

Movie A generates greater absolute profit.

Movie B generates much greater return relative to its investment.

Therefore:

Profit → Absolute financial performance

ROI → Investment efficiency

## Business Interpretation

The financial analysis shows that movie success should be evaluated
using multiple financial metrics rather than a single measure.

### Revenue

Gross revenue indicates the ability of a movie to generate sales
or box-office revenue.

### Profit

Profit measures the absolute financial gain after accounting for
the production budget.

### ROI

ROI measures how efficiently the production investment generated
profit.

### Budget

Higher budgets represent greater financial investment and therefore
greater financial exposure.

## Key Business Principle

A high-budget movie may generate high revenue and high absolute
profit but may not achieve the highest ROI.

Conversely, a low-budget movie can generate an exceptionally high
ROI even when its absolute profit is lower.

Therefore, investment decisions should consider both:

Profit + ROI

- The dataset represents only the movies available in the source
  dataset.
- Missing budget and gross values affect some financial analyses.
- Historical performance does not guarantee future success.
- Gross revenue does not represent complete studio profitability.
- Marketing costs, distribution costs and other expenses are not
  fully represented.
- ROI calculations depend on the availability and quality of budget
  and gross revenue data.

Therefore, the findings should be interpreted as analytical
insights rather than definitive predictions.
